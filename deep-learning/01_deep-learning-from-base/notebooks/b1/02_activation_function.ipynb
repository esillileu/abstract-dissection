# activation function
- 비선형성 부여
    - step
        - $y = \begin{cases}1 (x> 0) \\ 0 (x\leq 0)\end{cases}$
    - sigmoid
        - $y = \frac{1}{1+e^{-x}}$ 
    - relu
        - $y = \begin{cases}x (x > 0) \\ 0 (x \leq 0)\end{cases}$
    - softmax - 출력층
        - $\sigma(z)_i = \frac{e^{z_i}}{\sum^K_{j=1}e^{z_j}} = \frac{e^{z_i+c}}{\sum^K_{j=1}e^{z_j+c}} $


```python
# step function
def step(x):
    return np.where(x > 0, 1, 0)


print_function_graph(step, (-5, 5))
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


print_function_graph(sigmoid, (-5, 5))
```

```python
def relu(x):
    return np.maximum(0, x)


print_function_graph(relu, (-5, 5))
```

```python
def identity(x):
    return x


print_function_graph_3d(identity, (-5, 5), (-5, 5), output_index=1)
```

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)


print_function_graph_3d(softmax, (-5, 5), (-5, 5), output_index=1)
```
