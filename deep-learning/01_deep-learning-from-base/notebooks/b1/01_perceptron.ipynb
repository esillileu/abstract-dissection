# 퍼셉트론
$$
y = \begin{cases}0(b+w_1x_1+w_2x_2 \leq 0)\\1(b+w_1x_1+w_2x_2 > 0)\end{cases} = h(b+w_1x_1+w_2x_2) \\
h(x) = \begin{cases}0 (x \leq 0) \\ 1(x > 0)\end{cases}
$$
- $w$ : 각 변수의 영향력
- $b$ : 각 변수의 활성 기준 ($-\theta$)
- $h(x)$ : 활성화함수
- $b+w_1x_1+w_2x_2 =0$ 에 의한 경계선 생성 

```python
def AND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.7
    return 1 if np.sum(w * x) + b > 0 else 0


print_truth_table(AND)
```

```python
def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([-0.5, -0.5])
    b = 0.7
    return 1 if np.sum(w * x) + b > 0 else 0


print_truth_table(NAND)
```

```python
def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = -0.2
    return 1 if np.sum(w * x) + b > 0 else 0


print_truth_table(OR)
```

```python
# XOR(x1, x2) = AND(NAND(x1, x2), OR(x1, x2))
def XOR(x1, x2):
    s1 = NAND(x1, x2)
    s2 = OR(x1, x2)
    y = AND(s1, s2)
    return y


print_truth_table(XOR)
```
