```python
import numpy as np

import func as F
from datasets import mnist
from b1.utils import show_np_image
```

```python
def get_test_data():
    _, (x_test, t_test) = mnist.load_mnist(flatten=True, normalize=True)
    return (x_test, t_test)
```

# Neural network
- 다층 퍼셉트론
- 퍼셉트론 별 분리
- n층 nn에 대해 $$l_i(z) = h(W_i z  +b_i) \\ o(z) = \sigma(W_n z +b_n) \\ y = (o \circ l_{n-1}\circ l_{n-2} \cdots \circ l_{1})(x)$$
- $W_i$
    - 행 : 다음 층의 특정 뉴런에 대한 모든 연결
    - 열 : 현재 층의 특정 뉴런의 대한 모든 연결
    - ${W_i}_{(j, k)}$ : 다음층의 j 뉴런과 현재 층 k 뉴런 연결의 가중치 

```python
def init_network(): # 784, 50, 100, 10
    network = {}
    network["W1"] = np.random.rand(784, 50)
    network["b1"] = np.random.rand(50)
    network["W2"] = np.random.rand(50, 100)
    network["b2"] = np.random.rand(100)
    network["W3"] = np.random.rand(100, 10)
    network["b3"] = np.random.rand(10)

    return network
```

```python
def forward(network, x):
    W1, W2, W3 = network["W1"], network["W2"], network["W3"]
    b1, b2, b3 = network["b1"], network["b2"], network["b3"]

    a1 = np.dot(x, W1) + b1
    z1 = F.sigmoid(a1)
    a2 = np.dot(z1, W2) + b2
    z2 = F.sigmoid(a2)
    a3 = np.dot(z2, W3) + b3
    y = F.softmax(a3)

    return y
```

```python
network = init_network()
x = np.array([1.0, 0.5])
y = forward(network, x)
print(y)
```

## 배치 처리
- 행렬곱 연산에 대해 첫차원을 늘리는 것은 여러 값에 대해 동일한 계산을 적용하는 것과 같은 효과를 만들어냄


```python
x, t = get_test_data()
network = init_network()

batch_size = 100
accuracy_cnt = 0
for i in range(0, len(x), batch_size):
    x_batch = x[i:i+batch_size]
    t_batch = t[i:i+batch_size]

    y = forward(network, x_batch)
    p = np.argmax(y, axis=1)
    accuracy_cnt += np.sum(p == t_batch)

print("Accuracy:" + str(float(accuracy_cnt) / len(x)))
```
